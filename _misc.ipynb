```python
import pandas as pd

bb = pd.read_csv("synthemol/resources/real/building_blocks.csv")
preds = pd.read_csv("/compchem/arc/users/dvik/repos/SyntheMol/delqsar_real_preds.csv")

# Find prediction column (assume single prediction column besides smiles)
pred_cols = [c for c in preds.columns if c.lower() not in {"smiles"}]
if len(pred_cols) != 1:
    raise ValueError(f"Expected 1 prediction column, found {pred_cols}")
pred_col = pred_cols[0]

# Merge on smiles
out = bb.merge(preds[["smiles", pred_col]], on="smiles", how="left")
out = out.rename(columns={pred_col: "bioactivity"})

out.to_csv("synthemol/resources/real/delqsar_exp_building_blocks.csv", index=False)
print("Wrote:", "synthemol/resources/real/delqsar_exp_building_blocks.csv")
```
